```python
import allantools
import numpy as np
import matplotlib.pyplot as plt
```

# Using AllanTools

```python
N = 10000 # Num of points
t = np.logspace(0, 3, 50) # tau values from 1 to 1000
```

```python
t
```

```python
# Generate white noise
freq_white = allantools.noise.white(N)
freq_pink = allantools.noise.pink(N)
freq_rw = allantools.noise.brown(N)
```

```python
# Phases of each frequency noise
# Generate white noise
phase_white = np.cumsum(freq_white)
phase_pink = np.cumsum(freq_pink)
phase_rw = np.cumsum(freq_rw)
```

```python
f, (ax1, ax2, ax3) = plt.subplots(3, sharex=True)
ax1.set_title('Types of noises')
ax1.plot(freq_white)
ax2.plot(freq_pink)
ax3.plot(freq_rw)
f.subplots_adjust(hspace=0)
plt.setp([a.get_xticklabels() for a in f.axes[:-1]], visible=False)
```

```python
def allandev(freq):
    (taus, adevs, errors, ns) = allantools.adev(freq, data_type='freq', taus = t)
    return taus, adevs, errors, ns

def allanoadev(freq):
    (taus, adevs, errors, ns) = allantools.oadev(freq, data_type='freq', taus = t)
    return taus, adevs, errors, ns

def allanmdev(freq):
    (taus, adevs, errors, ns) = allantools.mdev(freq, data_type='freq', taus = t)
    return taus, adevs, errors, ns

def ohdev(freq):
    (taus, adevs, errors, ns) = allantools.ohdev(freq, data_type='freq', taus = t)
    return taus, adevs, errors, ns
```

```python
f, (ax1, ax2, ax3) = plt.subplots(3, sharex=True, sharey=True)
ax1.set_title('Allan Deviation')
ax1.loglog(allandev(freq_white)[0],allandev(freq_white)[1])
ax2.loglog(allandev(freq_pink)[0],allandev(freq_pink)[1])
ax3.loglog(allandev(freq_rw)[0],allandev(freq_rw)[1])
f.subplots_adjust(hspace=0)
plt.setp([a.get_xticklabels() for a in f.axes[:-1]], visible=False)
```

```python
f, (ax1, ax2, ax3) = plt.subplots(3, sharex=True, sharey=True)
ax1.set_title('Overlapping Allan Deviation')
ax1.loglog(allanoadev(freq_white)[0],allanoadev(freq_white)[1])
ax2.loglog(allanoadev(freq_pink)[0],allanoadev(freq_pink)[1])
ax3.loglog(allanoadev(freq_rw)[0],allanoadev(freq_rw)[1])
f.subplots_adjust(hspace=0)
plt.setp([a.get_xticklabels() for a in f.axes[:-1]], visible=False)
```

```python
f, (ax1, ax2, ax3) = plt.subplots(3, sharex=True, sharey=True)
ax1.set_title('Modified Allan Deviation')
ax1.loglog(allanmdev(freq_white)[0],allanmdev(freq_white)[1])
ax2.loglog(allanmdev(freq_pink)[0],allanmdev(freq_pink)[1])
ax3.loglog(allanmdev(freq_rw)[0],allanmdev(freq_rw)[1])
f.subplots_adjust(hspace=0)
plt.setp([a.get_xticklabels() for a in f.axes[:-1]], visible=False)
```

From Allan's paper:
It has been shown that if "flicker noise” frequency modulation is present ona signal from a standard, some significant problems arise,such as the logarithmic divergence of the standard deviation of the frequency fluctuations as the number of samples taken increases, and also the inability to define precisely the time average frequency.

```python
f, (ax1, ax2, ax3) = plt.subplots(3, sharex=True, sharey=True)
ax1.set_title('Overlapping Hadamard Deviation')
ax1.loglog(ohdev(freq_white)[0],ohdev(freq_white)[1])
ax2.loglog(ohdev(freq_pink)[0],ohdev(freq_pink)[1])
ax3.loglog(ohdev(freq_rw)[0],ohdev(freq_rw)[1])
f.subplots_adjust(hspace=0)
plt.setp([a.get_xticklabels() for a in f.axes[:-1]], visible=False)
```

# Self-Function

For some reason, tau yg direturn dari allantools lebih kecil dan rounded juga

```python
tau_values
```

```python
def adev_self(phase_data, tau):
    allan_variance = []

    for t in tau:
        # Calculate the Allan Variance for the current tau
        sum_variance = 0.0
        
        for i in range(N-2):
            sum_variance += (phase_data[i+2] - 2 * phase_data[i+1] + phase_data[i])**2
        
        allan_variance.append(sum_variance / (2*(N-2)*t**2))

    allandev = np.sqrt(allan_variance)
    return allandev
```

```python
adev_self(phase_white, tau_values)
```

```python
len(adev_self(phase_white,tau_values))
```

```python
allandev(freq_white)[1]
```

```python
plt.title('Comparasion of Allan Deviation')
plt.loglog(tau_values,allandev(freq_white)[1], label='allantools')
plt.loglog(tau_values,adev_self(phase_white,tau_values), label='self')
plt.legend()
```

```python
def oadev_self(phase_data,tau,m):
    over_allan_var = []

    for t in tau:
        sum_ov_var = 0.0
        
        for i in range(N-2*m):
            sum_ov_var += (phase_data[i+2*m] - 2 * phase_data[i+m] + phase_data[i])**2
        
        over_allan_var.append(sum_ov_var / (2*(N-2*m)*t**2))

    oallandev = np.sqrt(over_allan_var)
    return oallandev
```

```python
oadev_self(phase_white,tau_values,3)
```

```python
plt.title('Comparasion of Overlapping Allan Deviation')
plt.loglog(tau_values,allanoadev(freq_white)[1], label='allantools')
plt.loglog(tau_values,oadev_self(phase_white,tau_values,3), label='self')
plt.legend()
```

```python
def allandev_phase(phase):
    (taus, adevs, errors, ns) = allantools.adev(phase, data_type='phase', taus = t)
    return taus, adevs, errors, ns
```

```python
plt.title('Comparasion of Allan Deviation with phase data')
plt.loglog(tau_values,allandev_phase(phase_white)[1], label='allantools')
plt.loglog(tau_values,adev_self(phase_white,tau_values), label='self')
plt.legend()
```
