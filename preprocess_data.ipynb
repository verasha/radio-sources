```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import math
```

```python
def load_data(data_txt):
    cols = ['MJD','R.A. (mas)', 'Dec. (mas)', 'Err. R.A. (mas)', 'Err. Dec. (mas)', 'Corr.', 'Del.']
    cols_widths = [(1,9),(10,26),(27,46),(47,56),(57,66),(67,74),(75,79)]
    df = pd.read_fwf(data_txt, colspecs=cols_widths, skiprows=4, header=None, names=cols)    
    return df
```

```python
df_18 = load_data('1803+784_dem.txt')
df_18
```

```python
for col in df_18.columns:
    if df_18[col].dtype == 'object':
        try:
            df_18[col] = df_18[col].astype(float)
        except ValueError as e:
            print(f"{e}")
```

```python
def convert_value(value):
    try:
        return float(value)
    except ValueError:
        return None
```

```python
def preprocess_data(df):
    df_new = df.apply(pd.to_numeric, errors='coerce').dropna()
            
    return df_new
```

```python
df_18_new = preprocess_data(df_18)
df_18_new.dtypes
```
