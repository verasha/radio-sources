```python
import numpy as np
import statistics
```

```python
# NIST Weighted Variance

# Sample data
y = np.array([2,3,5,7,11,13,17,19,23])

# Sample weights
w = np.array([1,1,0,0,4,1,2,1,0])
```

```python
# WMA of data
ywbar = np.average(y,weights=w)
ywbar
```

```python
# Nprime = number of non zero weights
Np = np.count_nonzero(w)

# N = length of data
N = len(y)
```

$$ \begin{equation}
s_w^2 = \frac{\sum_{i=1}^N w_i (x_i-\bar{x}_w)^2}{(1-\frac{1}{N'})\sum_{i=1}^N w_i}
\end{equation}$$

```python
sum(w[i]*(y[i]-ywbar)**2 for i in range(N))/((1-1/Np)*sum(w[i] for i in range(N)))
```

```python
statistics.variance(y)
```

```python
def wvar(y,w):
    # Length of data
    N = len(y)
    # N prime = number of nonzero weights
    Np = np.count_nonzero(w)
    # Weighted mean average 
    ywbar = np.average(y,weights=w)
    # Weighted variance
    return sum(w[i]*(y[i]-ywbar)**2 for i in range(N))/((1-1/Np)*sum(w[i] for i in range(N)))
```

```python
wvar(y,w)
```
