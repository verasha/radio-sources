```python
import numpy as np
import matplotlib.pyplot as plt
import allantools 
import pandas as pd 
from datetime import date, timedelta
from astropy.time import Time
```

```python
def absln(x):
    if x == 0:
        return 0
    else:
        return np.log(np.abs(x))
```

```python
absln(0)
```

```python
def s_w(t,alpha):
    
    values = {2:-np.abs(t), \
              1:t**2*absln(t), \
              0:np.abs(t)**3, \
              -1:-t**4*absln(t), \
              -2:-np.abs(t)**5, \
              -3:t**6*absln(t), \
              4: np.abs(t)**7}
    
    if alpha in values.keys():
        return values[alpha]
    else:
        print('Return a value depending on the noise type')
```

```python
s_w(5,4)
```

```python
def s_x(t,F,alpha):
    return F**2*(2*s_w(t,alpha)-s_w(t-1/F,alpha)-s_w(t+1/F,alpha))
```

```python
s_x(1,1,1)
```

```python
def s_z(t,F,alpha):
    return 6*s_x(t,F,alpha)-4*s_x(t-1,F,alpha)-4*s_x(t+1,F,alpha)+s_x(t-2,F,alpha)+s_x(t+2,F,alpha)
```
