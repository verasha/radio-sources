```python
import numpy as np
import matplotlib.pyplot as plt
import allantools 
import pandas as pd 
from datetime import date, timedelta
from astropy.time import Time
```

# Generate white noise data with error

```python
N = 1000 # Num of points
```

```python
# Generate noise
freq_white = allantools.noise.white(N)
```

```python
sdate = date(2021,1,25)   # start date
edate = date(2023,10,22)   # end date
```

```python
# Generate dates
dates = pd.date_range(sdate,edate-timedelta(days=1),freq='d').to_julian_date()
dates.values
```

```python
# Convert to astropy Time object
t = Time(dates.values, format='jd')
t
```

```python
# Modified julian dy
t_mjd = t.mjd
t_mjd
```

```python
# Besselian year
t_by = t.byear
t_by
```

```python
# Generate random error
err = np.random.uniform(size=1000)
err
```

```python
df = pd.DataFrame({'Data':freq_white,'Error':err,'Julian':t,'MJD':t_mjd,'Besselian':t_by})
df
```

```python
plt.figure(figsize=(12,6))
plt.title('White Noise')
plt.errorbar(df['Besselian'], df['Data'],yerr=df['Error'], fmt='o', ms=1, mec='black', ecolor='lightgrey', elinewidth=1)
plt.xlabel('Year')
plt.ylabel('Freq')
plt.grid()
```

# Select partial data

```python
new_df = df.sample(700)
new_df
```

```python
plt.figure(figsize=(12,6))
plt.title('Partial White Noise')
plt.errorbar(new_df['Besselian'], new_df['Data'],yerr=new_df['Error'], fmt='o', ms=1, mec='black', ecolor='lightgrey', elinewidth=1)
plt.xlabel('Year')
plt.ylabel('Freq')
plt.grid()
```

# Find missing data

```python
# Running windows 
# check if there are any missing windows
m = 3
left = 0 
right = 0
D = 0
data_sum = 0

while right <= N:
    data_sum = np.sum(new_df['Data'].values[left:right])
    
    if data_sum == 0:
        D +=1
    left += 1
    right += 1
```
