```python
import numpy as np
import pandas as pd
```

```python
# Load radio coordinate time series data
radiodata = pd.read_fwf('opa2023s.txt', header=None)
```

```python
radiodata.head(5)
```

```python
radiodata['OBS'] = radiodata[16].astype(str) + ' ' + radiodata[17] + ' ' + radiodata[18].astype(str)
radiodata = radiodata.drop(radiodata.columns[[0,3,5,7,9,11,13,15,16,17,18]], axis=1)
radiodata.head(5)
```

```python
radiodata = radiodata.rename(columns={1: "SOU_LCO", 2: "DATETIME", 4: "EPOCH", 6: "RA", 8: "RA ERR", 10: "DEC", 12: "DEC ERR", 14: "CORR"})
radiodata.head(5)
```

Ekstrak data kumpulan epoch, RA, dan Dec dari setiap quasar berdasarkan nama quasarnya

```python
sourcedata = {}
for source, df_source in radiodata.groupby('SOU_LCO'):
    sourcedata[str(source)] = df_source
```
