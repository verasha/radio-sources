## Allan deviation the usual way

```python
def allanoadev(y,rate):
        
    # === Parameters ===
    
    N = len(y) # Ganti ke akhir-awal observasi
    # Nyoba tambahkan kondisi / algoritma ada missing windows 
    # Nyoba dulu di white / pink noise and random walk
    # m array where 3 < m < N/3 
    m = np.arange(3,1+round(N//3),1)
    # Take tau0 as 1/rate
    tau0 = 1/rate
    # steps
    step = 1
    
    # Empty variance array
    allanvar = []
    # Empty tau values array 
    tau = []

    # Iterate over values of m
    for k in range(len(m)): 
        # M is the number of times we can move the averaging window from the beginning to the end
        M = N-m[k]+1

        # Calculate allan var 
        outersum = []
        for j in np.arange(0,M-2*m[k]+1,step):
            innersum = []
            for i in np.arange(j,j+m[k]-1,step):
                innersum.append(y[i+m[k]]-y[i]) #weighted mean, bikin error sendiri - dist gaussian
            outersum.append(0.5*(np.sum(innersum)/m[k])**2)

        # Calculate allan variance
        allanvar.append(np.sum(outersum)/(M-2*m[k]+1))
        
        # Calculate tau
        tau.append(tau0*m[k])

    # Calculate allan std dev
    allanstd = np.sqrt(allanvar)
        
    return tau, allanstd
```

```python
%%time
tau_white, allanoadev_white = allanoadev(freq_white,1)
```

```python
plt.loglog(tau_white,allanoadev_white)
```

## Allan Deviation with mean

```python
def allanoadev_opt(y,rate):
        
    # === Parameters ===
    
    N = len(y)
    # m array where 3 < m < N/3 
    m = np.arange(3,1+round(N//3),1)
    # Take tau0 as 1/rate
    tau0 = 1/rate
    # steps
    step = 1
    
    # Empty variance array
    allanvar = []
    # Empty tau values array 
    tau = []

    # Iterate over values of m
    for k in range(len(m)): 
        kernels = np.ones(m[k])/m[k]
        ybar = np.convolve(y,kernels,'valid')

        # M is the number of times we can move the averaging window from the beginning to the end
        M = len(ybar)

        # Calculate allan var 
        outersum = []
        for j in np.arange(0,M-2*m[k]+1,step):
            innersum = []
            for i in np.arange(j,j+m[k]-1,step):
                innersum.append(ybar[i+m[k]]-ybar[i])
            outersum.append(0.5*(np.sum(innersum)/m[k])**2)

        # Calculate allan variance
        allanvar.append(np.sum(outersum)/(M-2*m[k]+1))
        
        # Calculate tau
        tau.append(tau0*m[k])

    # Calculate allan std dev
    allanstd = np.sqrt(allanvar)
        
    return tau, allanstd
```

```python
%%time
tau_white_opt, allanstd_white_opt = allanoadev_opt(freq_white,1)
```

```python
plt.loglog(tau_white_opt,allanstd_white_opt)
```

## Allan deviation with weighted mean

```python
def allanoadev_new(y,dates,error,rate):
        
    # === Parameters ===
    
    # Length of observation
    N = int(round((dates[-1] - dates[0])))
    
    # Nyoba tambahkan kondisi / algoritma ada missing windows 
    
    # m array where 3 < m < N/3 
    m = np.arange(3,1+round(N//3),1)
    # Take tau0 as 1/rate
    tau0 = 1/rate
    # steps
    step = 1
    
    # Empty variance array
    allanvar = []
    # Empty tau values array 
    tau = []
    
    # Create weights 
    weights_std = 1/(error**2)

    # Iterate over values of m
    for k in range(len(m)): 
        # Sliding windows 
        left = 0 
        right = m[k]
        ybar = []

        while right <= N:
            ybar.append(np.average(y[left:right], weights=weights_std[left:right]))
            left += step 
            right += step           

        # M is the number of times we can move the averaging window from the beginning to the end
        M = len(ybar)

        # Calculate allan var 
        outersum = []
        for j in np.arange(0,M-2*m[k]+1,step):
            innersum = []
            for i in np.arange(j,j+m[k]-1,step):
                innersum.append(ybar[i+m[k]]-ybar[i]) #weighted mean, bikin error sendiri - dist gaussian
            outersum.append(0.5*(np.sum(innersum)/m[k])**2)

        # Calculate allan variance
        allanvar.append(np.sum(outersum)/(M-2*m[k]+1))
        
        # Calculate tau
        tau.append(tau0*m[k])

    # Calculate allan std dev
    allanstd = np.sqrt(allanvar)
        
    return tau, allanstd
```

```python
%%time
tau_white_bar, allanoadev_white_bar = allanoadev_new(freq_white,t.mjd,err,1)
```

```python
plt.loglog(tau_white_bar,allanoadev_white_bar)
```

```python
plt.loglog(tau_white,allanoadev_white,label='fracfreq')
plt.loglog(tau_white_opt,allanstd_white_opt,label='usual mean')
plt.loglog(tau_white_bar,allanoadev_white_bar,label='weighted mean')
plt.title('Comparasion')
plt.legend()
```
